# Data Cleaning and Deduplication 

```python
def data_modification(data):
    lst = []

    for user in data:

            if user.get("attendance") == None:
                user["attendance"] = None

            if user.get("marks") == "thirty":
                user["marks"] = 30
            elif user.get("marks") == "seventy":
                user["marks"] = 70
        
            a = user.get("attendance")
            if type(user.get("attendance")) == type("a"):
                user["attendance"] = float(user.get("attendance"))
            if type(user.get("marks")) == type("a"):
                user["marks"] = float(user.get("marks"))
    
            a = user.get("marks")
            user["marks"] = float(a)
    
        
    return data
```

```python
data = data_modification(data)
print(data)
```

# Insights

```python
def insights(data):
    total_marks = 0
    below_forty = 0

    for user in data:
        x = user["marks"]
        total_marks = total_marks+x
        if user["marks"]<40:
            below_forty+=1
        

    
    print(f"Average marks of the students is {total_marks/len(data)} and percentage of below 40 is {(below_forty*100)/len(data)}")
```

```python
insights(data)     
```

# Recommendation

```python
def recommendation(data):
    
    lst1 = []

    for user in data:
        recommend = {}
        recommend["name"] = user["name"]
        if user["marks"]>=75:
            recommend["track"] = "Advanced"
        else:
            recommend["track"] = "Foundation"


        lst1.append(recommend)
    return lst1
```

```python
recommendation(data)
```
