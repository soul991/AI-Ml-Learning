# Data Cleaning, fixing Missing Values and fix Duplications

```python
import json
```

```python
# Load Data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data,"\n", type(data))
```

```python
# Clean and Structure The Data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        #Clean ratings
        raw_ratings = str(user["rating"]).strip().lower()
        if(raw_ratings in text_to_num):
            raw_ratings = text_to_num[raw_ratings]

        user["rating"] = raw_ratings

        #Missing values
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None


        #Deduplication
        if(user["name"].strip() in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
```

# Getting Meaningful Insights from Data

```python
def avg_rating_cal(data):
    sum = 0
    for user in data:
        sum += float(user["rating"])

    return (sum/len(data))
```

```python
avg_rating_cal(data)
```

```python
#User count with reviews less than 3
poor_rating = 0
for user in data:
    if(float(user["rating"])<3):
        poor_rating+=1
print(f"The percentage of poor rating is {((poor_rating)*100)/len(data)})")
```

# Building Recommendation feature

```python
def get_recommendation(data):
    recommendation = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        
        if(float(user["rating"])>=4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendation.append(curr_recomm)
    return recommendation
```

```python
get_recommendation(data)
```
