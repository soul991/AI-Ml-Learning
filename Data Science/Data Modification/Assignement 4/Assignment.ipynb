```python
with open("data.json", "r") as f:
    data = json.load(f)
```

# Data Cleaning

```python
def data_modification(data):



    for user in data:
        #fixing the steps values
        a = user.get("steps")
        lst = list(a)
        lst1 = []
        word = ""
        for ch in lst:
            if ch ==",":
                continue
            elif ch == "k":
                word+="000"
            else:
                word+=ch
        user["steps"] = int(word)

        #fixing names
        fixed_names = user["user"].strip()
        user["user"] = fixed_names

        #fixing missing values

        if user.get("weight") == None:
            user["weight"] = None


    return data
```

```python
data_modification(data)
```

# Insights

```python
def insights(data):
    total_steps = 0
    under_user = 0
    for user in data:
        x = user.get("steps")
        total_steps+=x

        if user.get("steps")<5000:
            under_user += 1

    print(f"The average steps of the users are {total_steps/len(data)} and total percentage of users under 5000 steps is/are {(under_user)*100/len(data)}")
```

```python
insights(data)
```
