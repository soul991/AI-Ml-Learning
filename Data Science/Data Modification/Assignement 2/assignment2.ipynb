```python
import json
```

```python
def load_json(filename):
    with open (filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_json("data.json")
print(data)
```

# Clean

```python
def data_clean(data):

    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    for user in data:
        raw_stars = str(user["stars"]).strip().lower()
        int_delivery_time, b = user["delivery_time"].split()
        a = int(int_delivery_time)
        if(raw_stars in text_to_num):
            raw_stars = text_to_num[raw_stars]
        else:
            raw_stars = float(raw_stars)

        if user.get("city") is None:
            user["city"]="Unknown"

        user["stars"] = raw_stars
        user["delivery_time"] = a


    return data
```

```python
data_clean(data)
```

# Deduplication

```python
def deduplication(data):
    unique = {}
    for user in data:
        unique[user["customer"].strip()] = user
    return list(unique.values())
```

```python
data = deduplication(data)
print(data)
```

# Insights

```python
def insights(data):
    total_time = 0
    customer_count = 0
    for user in data:
        time = user["delivery_time"]
        total_time+=time
        if user.get("stars")>=4:
            customer_count += 1
        
    print(f"The average delhivery time is {float((total_time)/len(data))} and customer with greater or equals to 4 stars is/are {customer_count}")
```

```python
insights(data)
```

# Recommendation

```python
def recommendation(data):
    lst = []
    
    
    for user in data:
        curr_recomm = {}
        curr_recomm["customer"] = user["customer"]
        if user.get("stars")>=4:
            curr_recomm["plan"] = "Premium"
        else:
            curr_recomm["plan"] = "Standard"

        lst.append(curr_recomm)

    return lst
        
```

```python
recommendation(data)
```
